## Part 3: Bar Graphs — Crime Data Visualization (2022–2023)

In this section, you will analyze real-world crime data from 2022 to 2023, sourced from the **FBI"s Crime Data** as reported by the Brennan Center. According to the Brennan Center’s analysis, most crimes, including murder, saw decreases in 2023.

Source: [FBI Data Confirms Drop in Most Crimes in 2023, Especially Murders](https://www.brennancenter.org/our-work/analysis-opinion/fbi-data-confirms-drop-most-crimes-2023-especially-murders)

## Dataset

| **Offense** | **Offenses per 100,000 People (2023)** | **Change in Offenses per 100,000 People (2022–2023)** |
|---|---:|---:|
| Aggravated Assault | 264.1 | -3.3 |
| Burglary | 250.7 | -8.1 |
| Larceny | 1,347.2 | -4.9 |
| Motor Vehicle Theft | 318.7 | +12.0 |
| Murder | 5.7 | -12.0 |
| Rape (revised definition) | 38.0 | -9.8 |
| Robbery | 66.5 | -0.8 |

---

````python
# convert markdown table to Pandas DataFrame Version 2

text = """
| **Offense** | **Offenses per 100,000 People (2023)** | **Change in Offenses per 100,000 People (2022–2023)** |
|---|---:|---:|
| Aggravated Assault | 264.1 | -3.3 |
| Burglary | 250.7 | -8.1 |
| Larceny | 1,347.2 | -4.9 |
| Motor Vehicle Theft | 318.7 | +12.0 |
| Murder | 5.7 | -12.0 |
| Rape (revised definition) | 38.0 | -9.8 |
| Robbery | 66.5 | -0.8 |
"""

# lines is a list of strings, because "\n" is replaced with ","
# remove lead & trail whitespaces, create lines at "new line" char
lines = text.strip().split("\n")

# row 0 is header line; remove lead & trail pipes
header_line = lines[0].strip("|")

# source:
# https://stackoverflow.com/questions/77068488/how-to-efficiently-convert-a-markdown-table-to-a-dataframe-in-python$0
````

````python
#lines
````

````python
#type(lines)
````

````python
#print(header_line)
````

````python
# split the header_line by pipes & remove all asterisks
headers = [h.strip().strip("*") for h in header_line.split("|")]
````

````python
#headers
````

````python
# The data lines start from the third line (index 2)
data_lines = lines[2:] # because skip header and ---- line

data = []  # create empty list
# each line (from row in md table) is a string
for line in data_lines:
    # "d" is each iterable element (substring) in "line"
    # Order:
    # 1) remove lead & trail (outer) pipes
    # 2) split using "|" char as delimiter (create list of substrings),
      # split() automatically removes the delimiter "|", replaces with ","
    # 3) still inside loop, remove any whitespace chars around the substrings
    row = [d.strip() for d in line.strip("|").split("|")]
    # "data" is a list of rows
    data.append(row)

# create Pandas DataFrame with "data" list of substrings as rows
  # and "headers" as column names
df = pd.DataFrame(data, columns = headers)
````

````python
# convert numeric cols to appropriate types (strings to floats), remove commas from nums
df["Offenses per 100,000 People (2023)"] = df["Offenses per 100,000 People (2023)"].str.replace(",", "").astype(float)
df["Change in Offenses per 100,000 People (2022–2023)"] = df["Change in Offenses per 100,000 People (2022–2023)"].astype(float)
````

## Task 3C: Improve the Clarity of Your Bar Graph

Create **one improved version** of your graph from Task 3B.

You may improve it by changing things such as:
- colors,
- labels,
- gridlines,
- figure size,
- text size,
- or the ordering of the categories.

Your goal is to make the graph easier to understand quickly.

---

````python
import seaborn as sns

# create increase/ decrease column
df["Change Type"] = df["Percent change 2022 to 2023"].apply(
    lambda x: "Increase" if x >= 0 else "Decrease") # small, anonymouse ftn

# sort df by percent change
df_sorted_percent = df.sort_values(
    by = "Percent change 2022 to 2023", ascending = True)

plt.figure(figsize = (12, 8))
# create horizontal bar chart
ax = sns.barplot(  # matplotlib axes object where the plot is drawn
    x = "Percent change 2022 to 2023",
    y = "Offense",
    hue = "Change Type",  # hue is color change depending on condition
    data = df_sorted_percent,
    palette = {"Increase": "red", "Decrease": "green"},
    dodge = False)  # dodge: “auto” or bool.  When hue mapping is used,
# whether elements should be narrowed and shifted along the orient axis
# to eliminate overlap. If "auto", set to True when the orient variable
# is crossed with the categorical variable or False otherwise.

plt.xlabel("Percent Change (2022 to 2023)", size = 16, labelpad = 20)
plt.ylabel("Offense Type", size = 16)
plt.title(
    "Percent Change in Crime Rates (2022-2023) by Offense Type", size = 20)
plt.grid(True, linestyle = "--", color = "grey", alpha = 0.7)
plt.axvline(x = 0, color = "black", linewidth = 2) # make 0-line bolder
plt.xlim(left = -75, right = 15) # adjust x-axis limits
plt.legend(title = "Direction of Change", title_fontsize = 18,
           loc = "lower left", fontsize = 16)

# add percentage labels near each bar
for container in ax.containers:    # Got help with this part.
    for bar in container:          # Need to learn about ax and containers
        # Get the width of the bar (the x-value for horizontal bars)
        width = bar.get_width()
        # Determine the x position for the text based on whether the bar is pos or neg
        #x_pos = width + 0.5 if width >= 0 else width - 5 # Adjusted for closer labels
        x_pos = width + 0.1 if width >= 0 else width - 4 # Adjusted for closer labels
        # Add the text label
        ax.annotate(f"{width:.1f}%",
                    # Position at the end of the bar
                    xy = (width, bar.get_y() + bar.get_height() / 2),
                    # Offset text position
                    xytext = (x_pos, bar.get_y() + bar.get_height() / 2),
                    # Center vertically, align horizontally
                    va = "center", ha = "center",
                    color = "black", fontsize = 16)

plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
````

## Task 3D: Multi-Series Grouped Bar Chart

In this task, you will practice making a **multi-series grouped bar chart** using a simple fictional dataset.

Imagine three companies (**Company A, Company B, Company C**) that each sell three products (**Product 1, Product 2, Product 3**).

### Example Dataset

```python
companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]
```

### Instructions
- Plot a grouped bar chart with **3 bars per company**.
- Use one bar for each product.
- Add labels, a title, and a legend.
- Make sure the three products are easy to compare within each company.

---

````python
companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]
````

````python
# Generate indexes for the x-axis (equal to the length of num of categories)
x_indexes = np.arange(len(companies))
print(x_indexes)
# Plot the first set of bars, shifting slightly to the left for proper alignment
plt.bar(x_indexes - 0.25, product1, width=0.25, label='Product 1')

# Plot the second set of bars, centered on 0
plt.bar(x_indexes, product2, width=0.25, label='Product 2')

# Plot the second set of bars, shifting slightly to the right for proper alignment
plt.bar(x_indexes + 0.25, product3, width=0.25, label='Product 3')

# Customize the x-axis ticks by setting them to the company names
plt.xticks(ticks = x_indexes, labels = companies)

plt.legend()
plt.xlabel("Company Name", fontsize = 16, labelpad = 20)
plt.ylabel("Number of Products Sold", fontsize = 16, labelpad = 20)
plt.title("Product Sales", fontsize = 20)

plt.show()
````
