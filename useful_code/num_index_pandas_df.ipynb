### Demonstrating Numerical Indexing with `.iloc`

Let's see how to access data in your `df` DataFrame using numerical (positional) indexing with `.iloc`.

```python
# Display the first row using .iloc
print("First row (position 0) using .iloc:")
display(df.iloc[0])  # only row 0

# first three rows [incl, excl]
# df.iloc[0:3]  # row 0, 1, 2

# Display the 'Offense' column (position 0) using .iloc
print("\nFirst column (position 0) using .iloc:")
display(df.iloc[:, 0])  # every row of col 0

# first two columns
# df.iloc[:, 0:2]

# Display the value at row 2, column 1 (position 2, 1) using .iloc
print("\nValue at row position 2, column position 1 (Aggravated Assault offense rate):")
display(df.iloc[2, 1])
```
